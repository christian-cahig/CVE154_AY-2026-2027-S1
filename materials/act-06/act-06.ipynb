# Programming Activity 06. Truss fund baby

*Last updated 02 November 2026*

## Scenario

Consider [`./act-06.png`](./act-06.png).
The dimensions of the ideal-truss tower is parameterized by $D$.
You are to solve for the member forces developed under any given loading.

## Task

### Part 0. Formulations

Treat each joint $i$ to be subject to three external actions:
- a net horizontal force $H_{i}$,
- a net vertical force $V_{i}$,
  and
- a net moment $C_{i}$.

These net external actions are the resultants
of all external loads applied at
and, where applicable, support reactions
at $i$.

Derive a system of linear equations
from which the member forces can be solved
for any given set of loads (forces and moments) on the tower.
Arrange the equations in matrix-vector form,
$\boldsymbol{A} \boldsymbol{x} = \boldsymbol{b}$.
As such,
$\boldsymbol{A}$ is a 15-by-15 matrix of coefficients,
$\boldsymbol{x}$ is a 15-element vector of the member forces to be solved,
and
$\boldsymbol{b}$ is a 15-element vector of right-hand-side (RHS) values.
Adopt the convention where tensile members have positive forces.

You should be able to show that
$\boldsymbol{A}$ depends on the geometry of the truss,
while $\boldsymbol{b}$ depends on the net external actions.

### Part 1. Core

*Answers to these tasks appear in the core part of the submission script.*

Import only the following:

- Python's math module as `mt`,
- NumPy as `np`,
- NumPy's [linear algebra submodule](https://numpy.org/doc/stable/reference/routines.linalg.html) as `npla`.

Store to the variable `__AUTHOR__` a string `"Group N"`,
where `N` is the appropriate group number.

Define a function `build_Ab` that takes four keyword arguments.
- `D` takes the value for $D$.
- `Fh` takes a list of net horizontal force loads (in kN).
- `Fv` takes a list of net vertical force loads (in kN).
- `Mc` takes a list of net moment loads (in kN-m).

`D` defaults to 1.
`Fh`, `Fv`, and `Mc` default to an empty list.

The net loads act on the joints,
and so the elements of `Fh`, `Fv`, and `Mc`
correspond to joints $A$, $B$, $C$, and so on until $I$, in that order.
The net loads shall be in their respective standard positive directions
(*i.e.*, right, up, counterclockwise).

Note that these net force loads and net moment loads are *applied loads*,
and therefore do not include support reactions.
Computations of the support reactions is a part of `build_AB`.
Together, the net force loads, net moment loads, and support reactions
make up the external actions ($H_{i}$'s, $V_{i}$'s, and $C_{i}$'s) on the joints.

`build_Ab` returns two NumPy arrays representing $\boldsymbol{A}$ and $\boldsymbol{b}$,
in that order.

You may define helper functions
(say, a helper function for computing the support reactions).
