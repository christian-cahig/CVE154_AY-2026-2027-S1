## Scenario

A hallmark of the
[Bernoulli-Euler beam theory](https://en.wikipedia.org/wiki/Euler%E2%80%93Bernoulli_beam_theory)
is the linear ordinary differential equation

$$\frac{\text{d}^{4}y}{\text{d}x^{4}} = \frac{f \left(x\right)}{E \left(x\right) \cdot I \left(x\right)}$$

This provides a relation between
- the *(static) distributed loading* $f \left(x\right)$
  borne by a beam,
- the *(static) deflection* $y \left(x\right)$
  of the neutral axis of the beam
  in response to $f \left(x\right)$,
- the material property of the beam,
  represented by
  the *modulus of elasticity* $E \left(x\right)$,
  and
- the geometric property of the beam,
  represented by the *moment of inertia* $I \left(x\right)$ of its cross section,

where $x$ denotes a location along the length of the beam
(typically referred from one end of the beam).
Graphically, $y \left(x\right)$ is the *elastic curve*.
The term $E \left(x\right) \cdot I \left(x\right)$ is also referred to as *flexural rigidity*.

Where the beam material is assumed linear, elastic, homogenous, and isotropic,
the elastic modulus is a constant $E$.
Similarly, if the beam's cross section is constant,
the moment of inertia is a constant $I$.
When such assumptions hold,

$$EI \frac{\text{d}^{4}y}{\text{d}x^{4}} = f \left(x\right)$$

Consider [`./act-04.png`](./act-04.png).
The $L$-m long beam is subject to a distributed force that linearly varies
from an upward intensity of $w_{L}$ N/m at its left end
to a downward intensity of $w_{R}$ N/m at the right end.
Assume that the beam is supported at its ends by hinges,
and that the Bernoulli-Euler theory holds.

## Task

### Part 0. Formulations

For a given set of $w_{L}$, $w_{R}$, and $L$,
derive:

- the support reactions $A_{y}$ and $B_{y}$ exerted at $A$ and $B$, respectively,
- the *internal shear* $\nu \left(x\right)$,
- the *bending moment* $\mu \left(x\right)$,
- the *slope* $m \left(x\right)$,
  and
- the deflection $y \left(x\right)$.

Assume unit flexural rigidity.

### Part 1. Core

*Answers to these tasks appear in the core part of the submission script.*

Import only the following:

- Python's `math` as `mt`,
- SciPy `optimize` as `spo`,
  and
- NumPy as `np`.

Store to the variable `__AUTHOR__` a string `"Group N"`,
where `N` is the appropriate group number.

Define functions `Ay` and `By`
that compute the values of $A_{y}$ and $B_{y}$, respectively,
for any given $w_{L}$, $w_{R}$, and $L$.
`Ay` and `By` each takes three positional arguments:
the first for $w_{L}$,
the second for $w_{R}$,
and
the third for $L$.

Define a helper function `polyvalder` that evaluates the value and the derivative of any polynomial
$a_{0} + a_{1} s + a_{2} s^{2} + a_{3} s^{3} + \cdots + a_{n-1} s^{n-1} + a_{n} s^{n}$.
`polyvalder` has two positional arguments:

1. a list containing the coefficients $a_{0}$, $a_{1}$, and so on up to $a_{n}$, in that order;
2. the value $s$ at which the value and the derivative of polynomial are to be evaluated.

`polyvalder` returns a NumPy array containing the polynomial value and derivative, in that order.
Use neither Python's `**` operator nor `mt.pow` in computing the polynomial value and derivative.
Instead, use only `*`, `+`, and `for` loop.

Define a function `nu`
that computes $\nu \left(x\right)$ and $\nu^{\prime} \left(x\right)$ at any location $x$
for any given $w_{L}$, $w_{R}$, and $L$.
`nu` takes four positional arguments:
one each for $x$, $w_{L}$, $w_{R}$, and $L$,
in that order.
`nu` returns a NumPy array containing $\nu \left(x\right)$ and $\nu^{\prime} \left(x\right)$, in that order.
Assume dimensional homogeneity is maintained.

Define a function `mu`
that computes $\mu \left(x\right)$ and $\mu^{\prime} \left(x\right)$ at any location $x$
for any given $w_{L}$, $w_{R}$, and $L$.
`mu` has the same call signature as `nu`.
Assume dimensional homogeneity is maintained.

Define a function `m`
that computes $m \left(x\right)$ and $m^{\prime} \left(x\right)$ at any location $x$
for any given $w_{L}$, $w_{R}$, and $L$.
`m` has the same call signature as `mu`.
Assume dimensional homogeneity is maintained.

Define a function `y`
that computes $y \left(x\right)$ and $y^{\prime} \left(x\right)$ at any location $x$
for any given $w_{L}$, $w_{R}$, and $L$.
`y` has the same call signature as `m`.
Assume dimensional homogeneity is maintained.

`nu`, `mu`, `m`, and `y` must use `polyvalder`.

Do not define any other functions.
