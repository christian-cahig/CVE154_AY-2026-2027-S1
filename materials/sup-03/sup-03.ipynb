## Preliminaries

```python
import math as mt

import scipy.optimize as spo
import numpy as np

from matplotlib import pyplot as plt
```

## Example 1

$$
f\left(x\right) = \left\{
    \begin{array}{ll}
        \left(x-3\right) \left(x-2\right) \left(x-1\right), & x \leq 4 \\
        \left(x-6\right)^{-1}, & 4 < x < 8 \\
        \operatorname{sgn} \left(x - 9\right), & 8 \leq x
    \end{array}
\right.
$$

```python
def f(x):
    if x <= 4:
        return (x - 3) * (x - 2) * (x - 1)
    elif (4 < x) and (x < 8):
        return 1.0 / (x - 6)
    else:
        return float(np.sign(x - 9).item())
```

```python
eps = 1e-7
lb, ub = 8.0, 27.0
iter_budget = mt.ceil((mt.log(ub - lb) - mt.log(eps)) / mt.log(2))

print(f"Tolerance: {eps}")
print(f"Iter. budget: {iter_budget}")
print(f"Search interval: [{lb}, {ub}]")
print(f"f(x) at bounds: {f(lb)}, {f(ub)}")
```

```python
p_bs, p_bs_info = spo.bisect(f, lb, ub, xtol = eps, rtol = eps, maxiter = iter_budget, disp = False, full_output = True)

print(p_bs, p_bs_info, sep="\n")
print(f"|{f(p_bs)}| <= {eps}? {abs(f(p_bs)) <= eps}")
```

## Example 2

$$
f \left(x\right) = \left( x^{2} - 1 \right) e^{-x^{2}} = \left( x^{2} - 1 \right) \operatorname{exp} \left(-x^{2}\right)
$$

$$
f^{\prime} \left(x\right) = 2x \left(2 - x^{2}\right) e^{-x^{2}}
$$

Properties:

- roots at $x = \pm 1$
- local extrema at $x = 0, \pm \sqrt{2}$
- asymptotically flat tails for $|x| > 2$

Initial guesses to consider:

- $x_{0} = 0.8$ for Newton-Raphson, $x_{0}, x_{1} = 0.7, 0.8$ for secant
- $x_{0} = -0.8$ for Newton-Raphson, $x_{0}, x_{1} = -0.7, 00.8$ for secant
- $x_{0} = 0.05$
- $x_{0} = 2.50$ for Newton-Raphson
- $x_{0} = 0.00$ for Newton-Raphson
- $x_{0} = \pm \sqrt{2}$ for Newton-Raphson

```python
def f(x):
    return ((x**2) - 1) * mt.exp(-(x**2))

def df(x):
    return (2 * x) * (2 - (x**2)) * mt.exp(-(x**2))
```

```python
eps = 1e-7
x_init = 0.8
iter_budget = 100

print(f"Tolerance: {eps}")
print(f"Iter. budget: {iter_budget}")
print(f"Search from: {x_init}")
print(f"f(x), f'(x) at start: {f(x_init)}, {df(x_init)}")
```

```python
p_nr, p_nr_info = spo.newton(f, x_init, tol = eps, rtol = eps, maxiter = iter_budget, disp = False, full_output = True)

print(p_nr, p_nr_info, sep="\n")
print(f"|{f(p_nr)}| <= {eps}? {abs(f(p_nr)) <= eps}")
```
