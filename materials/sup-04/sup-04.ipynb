## Preliminaries

```python
import numpy as np
import numpy.polynomial as npp
```

## Example

Consider the polynomial:
$$
\left(x - 5\right)^4
$$

Evaluate its value and its derivative.
Find (numerically) its roots, and evaluate it at those roots.

```python
gnome = npp.Polynomial([625, -500, 150, -20, 1])
d_gnome = gnome.deriv(1)
gnome_roots = gnome.roots()
```

```python
gnome
```

```python
d_gnome
```

```python
print(f"Value and first derivative of {gnome}")
for x in np.linspace(-11, 9, num=33):
    print(f"{' '*3}@ x = {x:.5f}:", f"{gnome(x):.7f};", f"{d_gnome(x):.7f}", sep="\t")
```

```python
print(gnome_roots)
```

```python
for r in gnome_roots:
    print(gnome(r))
```
