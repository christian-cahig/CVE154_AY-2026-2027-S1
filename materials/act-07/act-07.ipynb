# Programming Activity 07. Water you doing?

*Last updated 16 November 2026*

## Scenario

A portion of a wastewater treatment system includes
several tanks and a network of pipes.
Each tank receives an intake of pre-treated water
containing a treatment chemical at a certain concentration
(see [Schedule of tank intakes](#schedule-of-tank-intakes)).
A pipe conveys treated water from one tank to another.
Flow through the pipes is assumed to be at steady volumetric rates
(see [Schedule of pipe flows](#schedule-of-pipe-flows)).
Additionally, treated water may be withdrawn from each tank
(see [Schedule of tank withdrawals](#schedule-of-tank-withdrawals))
for other purposes.
In other words, the tank intakes are not from any of the pipes,
and the tank withdrawals do not go through any of the pipes.

Assue that uniform mixing occurs in each vessel.
Also assume that mass conservation holds,
*i.e.*,
the total mass of the chemical flowing into a tank
is equal to the total mass of the chemical flowing out of that tank.

## Task

You are to determine the concentration of the treatment chemical within the tanks,
for the given schedules of
[tank intakes](#schedule-of-tank-intakes),
[tank withdrawals](#schedule-of-tank-withdrawals),
and
[pipe flows](#schedule-of-pipe-flows).
Derive a system of linear equations
$\boldsymbol{A} \boldsymbol{x} = \boldsymbol{b}$,
where the unknown variables are the desired concentrations within the tanks.
Determine the determinant and condition number of the coefficient matrix $\boldsymbol{A}$.
Use Cramer's rule to solve for $\boldsymbol{x}$.

## Marking

A correct determinant earns 2 points.
A correct condition number earns 2 points.
A correct application of Cramer's rule earns 5 points.
Each correct concentration earns 2 points.
