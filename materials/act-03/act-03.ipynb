# Programming Activity 03. Peter Piper picked a peck of friction factors

*Last updated 24 September 2026*

## Scenario

The [Darcy-Weisbach model](https://en.wikipedia.org/wiki/Darcy%E2%80%93Weisbach_equation)
for flow of incompressible fluids in conduits
has a dimensionless term called the *friction factor*.

For turbulent flow in a round pipe,
the
*[Colebrook-White](https://doi.org/10.1680/ijoti.1939.14509)
[equation](https://youtu.be/dQw4w9WgXcQ)*
empirically relates the friction factor $f$
to characteristics of the pipe and of the flow:

$$\frac{1}{\sqrt{f}} = -2 \log \left( \frac{\varepsilon}{3.7 D} + \frac{2.51}{R \sqrt{f}} \right)$$

The terms inside the common logarithm are:

- $\varepsilon$, the absolute roughness of the pipe,
- $D$, the hydraulic diameter,
  and
- $R$, the Reynolds number.

If the pipe is completely filled and is circular,
$D$ is the inner diameter.
The dimensionless quantity $\varepsilon / D$ is also known as the *relative roughness* of the conduit.
[Lewis F. Moody](https://doi.org/10.1115/1.4018140)
reports that the Colebrook-White relation is valid
for $4000 \leq R \leq 10^{8}$.

The main objective of this Activity
is to numerically solve for the friction factor
from the Colebrook-White equation.

## Task

### Part 0. Formulations

From the Colebrook-White equation,
derive:
- a univariate function $z\left(x\right)$ of a variable $x$,
  and
- $z^{\prime} \left(x\right)$,
  *i.e.*, the derivative of $z$ with respect to $x$.

Given a relative roughness and a Reynolds number,
$z$ is zero when $x$ corresponds to the friction factor $f$.
In other words,
$z = 0$ means the left- and right-hand sides of the Colebrook-White equation
for a given pair of relative roughness and Reynolds number.
