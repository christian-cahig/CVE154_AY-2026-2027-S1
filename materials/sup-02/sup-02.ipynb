## Introduction

### SciPy routines

This material is primarily for you to learn how to use
the following univariate root-finding methods
as they are implemented in SciPy's
[`optimize`](https://docs.scipy.org/doc/scipy/tutorial/optimize.html#root-finding) package.

- bisection,
  via
  [`scipy.optimize.bisect`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.bisect.html)
- Newton-Raphson,
  via
  [`scipy.optimize.newton`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.newton.html)
- secant,
  via
  [`scipy.optimize.newton`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.newton.html)
- TOMS Algorithm 748,
  via
  [`scipy.optimize.toms748`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.toms748.html)


Bisection and TOMS Algorithm 748 are *bracketing methods*
because they search for a root within a user-provided interval
that is guaranteed to
["bracket a root"](https://en.wikipedia.org/wiki/Root-finding_algorithm#Bracketing_methods).
The rest are considered *open methods*.

### lambda functions

Another aim of this material is to introduce
Python's [lambda functions](https://www.w3schools.com/python/python_lambda.asp).
In particular, we will be concerned with using lambda functions
to define a function that is of a "narrower capability"
than that of another.
As an example, consider the following snippet.

<!-- ```python

````python
def full_func(a, b = 1, c = 2):
    return (a + b) / 2

B = 5; C = 3

reduced_func = lambda x : full_func(x, b = B, c = C)
````

For any numeric value stored to `x`,
`reduced_func(x)`
will return the same value as
`full_func(x, b = B, c = C)`.

````python
print(reduced_func(4) == full_func(4, b = B, c = C))
print(reduced_func(20) == full_func(20, b = B, c = C))
print(reduced_func(17) == full_func(17, b = B, c = C))
````

## Scenario

Consider the trough with semicircular cross section
in [Programming Activity 01](../act-01/).

The trough dimensions $d$ and $l$ are known *a priori*.
For a given water volume $V$,
you need to determine $h$
and the wetted perimeter $P$.

## Task

### Part 0

Derive a function $f\left(x\right)$
of a variable $x$,
whose root is the desired $h$,
*i.e.*,
$f\left(h\right) = 0$.

Express $P$ as a function of $h$.

Derive $f^{\prime}\left(x\right)$,
*i.e.*,
the derivative of $f$ with respect to $x$.
This will be necessary when using Newton-Raphson.

In other words, you are to go through two subproblems:
1. determining $h$ by solving for the root of $f\left(x\right)$,
   and
2. computing $P$ corresponding to that $h$.

### Part 1

Import only the following:

- `math` under the alias `mt`,
  and
- `scipy.optimize` under the alias `spo`.

As a convetion for this course,
SciPy's `optimize` should always be imported under the alias `spo`.
Thus, the root-finding routines shall be invoked as
`spo.bisect`, `spo.newton`, and `spo.toms748`.

Define functions `uptown_func`, `daft_func`, and `wetper`.

`uptown_func`
implements $f\left(x\right)$,
and takes the following input arguments.

- a required argument
  representing $x$
- an optional argument `d`
  representing $d$
  and having a default value of 7
- an optional argument `l`
  representing $l$
  and having a default value of 10
- an optional argument `V`
  representing $V$
  and having a default value of 17


`daft_func` computes
$f^{\prime}\left(x\right)$
and has the same input arguments as `uptown_func`.

`wetper` has two required required arguments:

1. the distance $h$,
   and
2. the diameter $d$,

and outputs the corresponding wetted perimeter $P$.

Realize that each of these functions outputs a single number.
Assume dimensional homogeneity is maintained.

### Part 2

In the synthesis part of the submission script,
define the following variables.

- `dia` stores the value for $d$.
- `len` stores the value for $l$.
- `vol` stores the value for $V$.
- `lb` and `ub` store the lower and upper ends, respectively,
  of an initial bracketing interval.
  These are used in bracketing methods.
- `x0` stores the initial guess of the root.
  This is not used in bracketing methods.
- `iter_budget` stores the iteration budget to be used for the methods.

You are to solve for $h$ when

- $d$ is 1 meter,
- $l$ is 3 meters,
  and
- $V$ is 1 cubic meters.

Use an initial bracketing interval of your choice.
For non-bracketing methods,
set the initial guess to be the midpoint of the initial bracketing interval.
Use an iteration budget of 1000.

### Part 3

Define lambda functions `f` and `df`
that are essentially `uptown_func` and `daft_func`,
respectively,
but with the optional arguments fixed to the values specified in Part 2.

Unless otherwise required,
use `f` and `df` as applicable in the succeeding Parts.
As such,

- `f` is to be passed to all routines
  as the Python function implementation
  of the mathematical function $f\left(x\right)$
  whose root must be found,
  and
- `df` is only used in the Newton-Raphson routine.

### Part 4

Estimate the root $h_{\text{bs}}$
of $f\left(x\right)$ using bisection method.
In calling
[`spo.bisect`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.bisect.html),

- pass `True` to the optional argument `full_output`,
  and
- pass `False` to the optional argument `disp`.

As such, `spo.bisect` will output two quantities:

1. the estimate of the desired root,
   and
2. a container of information on how the method ran,

which you must store to variables `h_bs` and `h_bs_info`, respectively.

Define variables `f_bs`, `p_bs`, and `df_bs`,
and store thereto the values of
$f\left(x\right)$, $P$, and $f^{\prime}\left(x\right)$ respectively,
corresponding to $h_{\text{bs}}$.

### Part 5

Estimate the root $h_{\text{nr}}$
of $f\left(x\right)$ using the Newton-Raphson method.
In calling
[`spo.newton`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.newton.html),

- pass `df` to the optional argument `fprime`,
- pass `True` to the optional argument `full_output`,
  and
- pass `False` to the optional argument `disp`.

`spo.newton` called thus yields two outputs:

1. the estimate of the desired root,
   and
2. a container of information on how the method ran,

which you must store to variables `h_nr` and `h_nr_info`, respectively.

Define variables `f_nr`, `p_nr`, and `df_nr`,
and store thereto the values of
$f\left(x\right)$, $P$, and $f^{\prime}\left(x\right)$ respectively,
corresponding to $h_{\text{nr}}$.

### Part 6

Estimate the root $h_{\text{sc}}$
of $f\left(x\right)$ using the secant method.
You will still use
[`spo.newton`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.newton.html)
but will no longer use the analytical derivative.
So,

- pass the lower end of the initial bracketing interval to the optional argument `x1`,
- pass `True` to the optional argument `full_output`,
  and
- pass `False` to the optional argument `disp`.

(The derivative at the first iteration is estimated `x0` and `x1`.
This is in contrast to the Newton-Raphson scheme
where the derivative at the first iteration is estimated
by passing `x0` to `df`.
Moreover,
just because the secant method requires two estimates from the start
does not mean it is a bracketing method.)

As in Part 7, there will be two output quantities:

1. the estimate of the desired root,
   and
2. a container of information on how the method ran,

which you must store to variables `h_sc` and `h_sc_info`, respectively.

Define variables `f_sc`, `p_sc`, and `df_sc`,
and store thereto the values of
$f\left(x\right)$, $P$, and $f^{\prime}\left(x\right)$ respectively,
corresponding to $h_{\text{sc}}$.

### Part 7

Estimate the root $h_{\text{to}}$
of $f\left(x\right)$ using TOMS Algorithm 748.
In calling
[`spo.toms748`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.toms748.html),

- pass `True` to the optional argument `full_output`,
  and
- pass `False` to the optional argument `disp`.

Similar to Part 4, there will be two outputs:

1. the estimate of the desired root,
   and
2. a container of information on how the method ran,

which you must store to variables `h_to` and `h_to_info`, respectively.

Define variables `f_to`, `p_to`, and `df_to`,
and store thereto the values of
$f\left(x\right)$, $P$, and $f^{\prime}\left(x\right)$ respectively,
corresponding to $h_{\text{to}}$.
