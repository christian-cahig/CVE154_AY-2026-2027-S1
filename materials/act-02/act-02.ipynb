# Programming Activity 02. Influence in the future

*Last updated 13 September 2026*

## Scenario

[`./act-02.png`](./act-02.png)
depicts a girder supporting a floor system.
A point load is on top of the *stringer*
modelled as a combination of *panels*.
*Floor beams* at the panel points
(*i.e.*, endpoints of the stringer panels)
transmit the load to the *girder*.

The load has a magnitude of $u$ kip,
and is located $x$ ft from the left end of the stringer.

## Task

### Part 1

Derive the following in terms of
$x$, $u$, $P$, and $H$.

- $A_{y}$, the vertical force developed at the roller support
- $E_{y}$, the vertical force developed at the fixed support
- $E_{m}$, the reaction moment developed at the fixed support
- $F_{y}$, the magnitude of the force developed at the hinge
- $\mu_{G}$, the bending moment developed at $G$

The hinge is to the right of panel point $B$
but to the left of panel point $C$.

Note that $G$ is on the girder
and is directly below the midpoint of panel $CD$.

Assume that
$A_{y}$, $E_{y}$, $E_{m}$, and $\mu_{G}$
are oriented according to standard positive directions.

### Part 2

Import only Python's [`math`](https://docs.python.org/3/library/math.html) under the alias `mt`.

Store to the variable `__AUTHOR__`
a string `"Group N"`,
where `N` is the appropriate group number.

Define five functions
`Ay`, `Ey`, `Em`, `Fy`, and `mu_G`
corresponding respectively to
$A_{y}$, $E_{y}$, $E_{m}$, $F_{y}$, and $\mu_{G}$.
Each of these has two required (positional) arguments:

1. the value for $u$, and
2. the value for $x$,

along with two optional (keyword) arguments:

- `P`
  which takes the value of $P$ and defaults to 10,
  and
- `H`
  which takes the value of $H$ and defaults to 15.

Maintain dimensional homogeneity.

### Part 3

Define the following variables
in the synthesis part of the submission script,
prior to the `print` statements.

- `load_mag` for storing a value for $u$
- `load_loc` for storing a value for $x$
- `panel_len` for storing a value for $P$
- `hinge_loc` for storing a value for $H$

You can assign values of your choice to these variables.

Do not modify any pre-written lines you will find in the synthesis part.
