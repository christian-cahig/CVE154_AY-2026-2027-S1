## Preliminaries

```python
import numpy as np
import numpy.linalg as npla
```

## Example 1

Solve for
$c_{1}$, $c_{2}$, $c_{3}$, $c_{4}$, and $c_{5}$
from the following equations.

$-8 c_{1} + 4 c_{2} = -80$

$8 c_{1} - 10 c_{2} + 2 c_{3} = 0$

$6 c_{2} - 11 c_{3} + 5 c_{4} = 0$

$3 c_{3} - 7 c_{4} + 4 c_{5} = 0$

$2 c_{4} - 4 c_{5} = -30$

### Forming the matrix-vector form $\boldsymbol{A} \boldsymbol{x} = \boldsymbol{b}$

$\boldsymbol{A}$ is the coefficient matrix.
$\boldsymbol{x}$ is the unknowns vector.
$\boldsymbol{x}$ is the RHS vector.

```python
A = np.array([
    [-8, 4, 0, 0, 0],
    [8, -10, 2, 0, 0],
    [0, 6, -11, 5, 0],
    [0, 0, 3, -7, 4],
    [0, 0, 0, 2, -4]
])
b = np.array([-80, 0, 0, 0, -30])
```

```python
print("Coefficient matrix:", A, sep="\n")
print("RHS vector:", b, sep="\n")
```

### Determinant, condition number, and inverse of $\boldsymbol{A}$

A linear system of equations has a solution if $\boldsymbol{A}$ is nonsingular,
which means that:
- the determinant of $\boldsymbol{A}$ is nonzero,
  and
- $\boldsymbol{A}$ has an inverse $\boldsymbol{A}^{-1}$
  for which $\boldsymbol{A} \boldsymbol{A}^{-1} = \boldsymbol{I}$,
  where $\boldsymbol{I}$ is the identity matrix.

The condition number of $\boldsymbol{A}$
is $\infty$ if it is singular,
and is a finite value otherwise.
The system is ill-conditioned
if $\boldsymbol{A}$ has a very large, finite condition number.

```python
print(f"Determinant of coefficient matrix: {npla.det(A)}")
print(f"Condition number of coefficient matrix: {npla.cond(A)}")

A_inv = npla.inv(A)
print("Inverse of coefficient matrix:", A_inv, sep="\n")
```

### Solving for the unknowns

```python
x = npla.solve(A, b)
print("Values of unknowns:", x, sep="\n")
```

### Verifying the solution

If $\hat{\boldsymbol{x}}$ is a solution,
then $\boldsymbol{A} \hat{\boldsymbol{x}}$ must be equal (or practically equal) to $\boldsymbol{b}$.
In other words,
$\boldsymbol{A} \hat{\boldsymbol{x}} - \boldsymbol{b}$ must be equal (or practically equal) to a vector of zeros.

```python
print(A @ x)
print(b)
print((A @ x) - b)
```

Equivalently,
$\hat{\boldsymbol{x}}$ must be equal (or practically equal) to
$\boldsymbol{A}^{-1} \boldsymbol{b}$.

```python
print(x)
print(A_inv @ b)
print(x - (A_inv @ b))
```

## Example 2

Solve the same system of equations as in Example 1,
but now use Cramer's rule.

In essence,
the $i$-th unknown $x_{i}$
is the ratio of
the determinants of $\boldsymbol{A}^{\left(i\right)}$
and of $\boldsymbol{A}$:

$$
x_{i} = \frac{\operatorname{det} \boldsymbol{A}^{\left(i\right)}}{\operatorname{det} \boldsymbol{A}}
$$

where $\boldsymbol{A}^{\left(i\right)}$ is $\boldsymbol{A}$
but whose $i$-th column is replaced with $b$.

```python
# Use the `A.copy()` so that `A1`, ..., `A5` are separate copies of -- and not merely pointing to -- `A`
A1 = A.copy(); A2 = A.copy(); A3 = A.copy(); A4 = A.copy(); A5 = A.copy()

A1[:,0] = b; A2[:,1] = b; A3[:,2] = b; A4[:,3] = b; A5[:,4] = b
```

```python
x = np.array([npla.det(A1), npla.det(A2), npla.det(A3), npla.det(A4), npla.det(A5)]) / npla.det(A)
print("Values of unknowns:", x, sep="\n")
```

```python
print(A @ x)
print(b)
print((A @ x) - b)

print(x)
print(A_inv @ b)
print(x - (A_inv @ b))
```
